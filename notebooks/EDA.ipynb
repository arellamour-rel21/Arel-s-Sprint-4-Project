```python
import pandas as pd
vehicles_df = pd.read_csv('../vehicles_us.csv')
vehicles_df.head()
```

```python
vehicles_df.info()
```

```python
vehicles_df['date_posted'] = pd.to_datetime(vehicles_df['date_posted'])
vehicles_df['is_4wd'] = vehicles_df['is_4wd'].fillna(0)
vehicles_df['is_4wd'] = vehicles_df['is_4wd'].astype(int)
vehicles_df['is_4wd'].value_counts()
vehicles_df['paint_color'] = vehicles_df['paint_color'].fillna('unknown')
vehicles_df['model_year'] = vehicles_df['model_year'].fillna(vehicles_df['model_year'].median())
vehicles_df['model_year'] = vehicles_df['model_year'].astype(int)
vehicles_df['odometer'] = vehicles_df['odometer'].fillna(vehicles_df['odometer'].median())
vehicles_df['odometer'] = vehicles_df['odometer'].astype(int)
vehicles_df['cylinders'] = vehicles_df['cylinders'].fillna(vehicles_df['cylinders'].mode()[0])
vehicles_df['cylinders'] = vehicles_df['cylinders'].astype(int)
vehicles_df['transmission'] = vehicles_df['transmission'].fillna('unknown')
vehicles_df['price'] = vehicles_df['price'].fillna(vehicles_df['price'].median())
vehicles_df['price'] = vehicles_df['price'].astype(int)

vehicles_df.info()
```

```python
import plotly.express as px
myear_fig = px.histogram(vehicles_df, x='model_year', nbins=20, title='Distribution of Model Year')
myear_fig
```

```python
price_grp = vehicles_df.groupby('model_year')['price'].mean().reset_index()
price_fig = px.scatter(price_grp, x='model_year', y='price', title='Average Price by Model Year')
price_fig
```

```python
odom_fig = px.scatter(vehicles_df, x='odometer', y='price', title='Odometer vs. Price')
odom_fig
```
